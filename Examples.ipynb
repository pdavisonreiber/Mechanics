# Mechanics DSL Quickstart

This notebook uses the mechanics DSL to solve one A-Level connected-particles problem.

Use plain assignment for unknowns, for example `a = unknown_acceleration()`. In notebooks, if direct source-line inference is not available, the DSL resolves names from the notebook namespace when you build equations, solve, or show diagrams.


## Example 1: Connected Particles on a Rough Plane

Two particles, `P` and `Q`, of masses 5 kg and 3 kg respectively, are connected by a light inextensible string. Particle `P` is pulled by a horizontal force of magnitude 40 N along a rough horizontal plane. Particle `P` experiences a frictional force of 10 N and particle `Q` experiences a frictional force of 6 N.

Find the acceleration of the particles and the tension in the string.


## Imports

```python
from mechanics import g, particle, system, unknown_acceleration, unknown_force
```

## Model

Take right as the positive horizontal direction. The string is light, so the tension has the same magnitude throughout. The string is inextensible, so both particles have the same acceleration `a`.


```python
P = particle(mass=5)
Q = particle(mass=3)

T = unknown_force()
a = unknown_acceleration()

P.force(horizontal=40)
P.force(horizontal=-10)
P.force(horizontal=-T)

Q.force(horizontal=T)
Q.force(horizontal=-6)

s = system(Q, P)
s.acceleration(horizontal=a)
```

## Force Diagrams

```python
s.diagrams()
```

## Equations

```python
s.equations()
```

## Solution

```python
s.solve()
```

## Exact Solution

```python
s.solve_exact()
```

## Modelling Assumptions

The string is modelled as light, so the tension is the same at both ends of the string. The string is modelled as inextensible, so particles `P` and `Q` have the same acceleration.


## Example 2: Scale-Pan Raised Vertically

A light scale-pan is attached to a vertical light inextensible string. The scale-pan carries two masses `A` and `B`. The mass of `A` is 400 g and the mass of `B` is 600 g. `A` rests on top of `B`.

The scale-pan is raised vertically with acceleration `0.5 m s^-2`.

Find the tension in the string, the force exerted on mass `B` by mass `A`, and the force exerted on mass `B` by the scale-pan.


## Model

Take upward as the positive vertical direction. Let `R` be the contact force between `A` and `B`, and let `T` be the upward force exerted by the scale-pan/string on `B`.

The string is light, so the tension is transmitted through the pan. The pan is light, so its own weight is ignored. The system is raised together, so both masses have acceleration `0.5 m s^-2` upward.


```python
A = particle(mass=0.4)
B = particle(mass=0.6)

T = unknown_force()
R = unknown_force()

A.force(vertical=R)
A.force(vertical=-0.4 * g)

B.force(vertical=T)
B.force(vertical=-R)
B.force(vertical=-0.6 * g)

scale_pan = system(A, B)
scale_pan.acceleration(vertical=0.5)
```

## Force Diagrams

```python
scale_pan.diagrams()
```

## Equations

```python
scale_pan.equations()
```

## Solution

```python
scale_pan.solve()
```

The tension in the string is `T`. The force exerted on mass `B` by mass `A` has magnitude `R` and acts downward. The force exerted on mass `B` by the scale-pan has magnitude `T` and acts upward.

## Exact Solution

```python
scale_pan.solve_exact()
```
