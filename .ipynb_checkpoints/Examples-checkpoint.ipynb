# Mechanics DSL Quickstart

This notebook shows how to use the SymPy-backed mechanics DSL for A-Level style force and acceleration problems.

In notebooks, use explicit names for unknowns, for example `unknown_acceleration("a")`. Jupyter does not always expose the source line in a way that lets the DSL infer names from assignments like `a = unknown_acceleration()`.

## Imports

```python
from mechanics import g, particle, system, unknown_acceleration, unknown_force
```

## Example 1: Vertical Motion

A particle of mass `5` has vertical forces `5` and `-3g`. Find its vertical acceleration `a`.

```python
A = particle(mass=5, name="A")
a = unknown_acceleration("a")

A.force(vertical=5)
A.force(vertical=-3 * g)

s = system(A)
s.acceleration(vertical=a)

s.show_equations()
```

```python
s.solve(a)
```

```python
s.show_diagrams()
```

## Example 2: Unknown Horizontal Force

A particle of mass `5` accelerates horizontally at `3`. Find the horizontal force `T`.

```python
A = particle(mass=5, name="A")
T = unknown_force("T")

A.force(horizontal=T)

s = system(A)
s.acceleration(horizontal=3)

s.show_equations()
```

```python
s.solve(T)
```

## Example 3: Matching the Original Syntax

This mirrors the syntax sketch, with the small notebook-specific change that unknowns and particles are named explicitly.

```python
A = particle(mass=5, name="A")
B = particle(mass=2, name="B")
T = unknown_force("T")
a = unknown_acceleration("a")

A.force(vertical=5)
A.force(vertical=-3 * g)
B.force(horizontal=-T)

s = system(A, B)
s.acceleration(horizontal=5)

s.solve(T)
```

```python
s.show_equations()
```

```python
s.show_diagrams()
```

```python
s.acceleration(vertical=a)
s.show_equations()
```

## Sign Convention

Signs are explicit. Choose positive horizontal and vertical directions for the problem, then enter opposing forces as negative values. For example, if upward is positive, a downward weight is written as `vertical=-5 * g`.
